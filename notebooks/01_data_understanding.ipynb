## 3. Detailed Data Inspection

In this section, we will inspect the data types, identify missing values, check for duplicate records, and calculate summary statistics.

This analysis will help us understand data quality and decide which preprocessing steps are required before training machine learning models.

```python
# Inspect columns and data types in both datasets.
for name, data in {"train": train, "test": test}.items():
    print(f"{name.title()} columns:", data.columns.tolist())
    data.info()
```

```python
# Count missing values in both datasets.
for name, data in {"train": train, "test": test}.items():
    print(f"{name.title()} missing values:\n", data.isna().sum(), "\n")
```

```python
# Calculate missing-value percentages for the training data.
missing_percentage = train.isna().mean().mul(100).round(2)
missing_percentage
```

```python
# Count duplicate rows in both datasets.
for name, data in {"train": train, "test": test}.items():
    print(f"Duplicate rows in {name}: {data.duplicated().sum()}")
```

```python
# Show summary statistics for numerical columns.
train.describe()
```

## 4. Categorical Data and Unusual Values

In this section, we will examine categorical variables to identify inconsistent labels, understand the distribution of product and outlet categories, and investigate unusual numerical values.

These findings will guide our data-cleaning and preprocessing decisions.

```python
# Compare fat-content categories in both datasets.
for name, data in {"train": train, "test": test}.items():
    print(f"{name.title()} fat content:\n", data["Item_Fat_Content"].value_counts(), "\n")
```

```python
# Count important categorical values.
for column in ["Outlet_Size", "Item_Type", "Outlet_Type"]:
    print(f"{column}:\n", train[column].value_counts(dropna=False), "\n")
```

```python
# Count unique products and outlets.
print("Unique products:", train["Item_Identifier"].nunique())
print("Unique outlets:", train["Outlet_Identifier"].nunique())
```

```python
# Count items with zero visibility.
zero_visibility = train["Item_Visibility"].eq(0).sum()
print("Zero visibility:", zero_visibility)
print("Percentage:", round(zero_visibility / len(train) * 100, 2), "%")
```

## 5. Data Cleaning

Data cleaning improves data consistency and prepares the dataset for machine learning. Based on the initial inspection, the main issues are inconsistent fat-content labels, missing values in `Item_Weight` and `Outlet_Size`, and zero values in `Item_Visibility`.

We will first standardize categorical labels and investigate outlet-size patterns. Missing-value imputation and other transformations will be handled carefully during preprocessing to avoid data leakage.

```python
# Create copies of the original datasets
train_clean = train.copy()
test_clean = test.copy()
```

```python
# Standardize fat-content labels
mapping = {
    "LF": "Low Fat",
    "low fat": "Low Fat",
    "reg": "Regular"
}

# Apply the same changes to both datasets
train_clean["Item_Fat_Content"] = train_clean["Item_Fat_Content"].replace(mapping)
test_clean["Item_Fat_Content"] = test_clean["Item_Fat_Content"].replace(mapping)

# Check the results
print(train_clean["Item_Fat_Content"].value_counts())
```
