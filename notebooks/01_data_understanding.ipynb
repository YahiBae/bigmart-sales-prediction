## 1. Project Introduction

The objective of this project is to develop a machine learning model that predicts product sales at different BigMart outlets using the provided historical dataset.

The project will involve understanding the dataset, handling missing values, performing exploratory data analysis (EDA), preprocessing features, training machine learning models, and evaluating their performance.

## 2. Dataset Understanding

The first step is to inspect the training and test datasets to understand their structure, columns, data types, number of records, and missing values.

The training dataset contains the target variable used to train the model. The test dataset will be inspected separately to understand which features are available for prediction.

```python
# Import the pandas library for working with tabular data
import pandas as pd

# Load the training and test datasets
train_path = "../data/raw/Train.csv"
test_path = "../data/raw/Test.csv"

train = pd.read_csv(train_path)
test = pd.read_csv(test_path)
```

```python
# Display the first five rows of the training dataset
train.head()
```

```python
# Display the first five rows of the test dataset
test.head()
```

```python
# Check the number of rows and columns in each dataset
print("Training dataset shape:", train.shape)
print("Test dataset shape:", test.shape)
```

## 3. Detailed Data Inspection

In this section, we will inspect the data types, identify missing values, check for duplicate records, and calculate summary statistics.

This analysis will help us understand data quality and decide which preprocessing steps are required before training machine learning models.

```python
# Display the column names in the training dataset
print("Training dataset columns:")
print(train.columns.tolist())

# Display information about the training dataset
train.info()
```

```python
# Display the structure of the test dataset
test.info()
```

```python
# Count missing values in each training dataset column
print("Missing values in training data:")
print(train.isnull().sum())

# Count missing values in each test dataset column
print("\nMissing values in test data:")
print(test.isnull().sum())
```

```python
# Calculate the percentage of missing values in each training column
missing_percentage = (train.isnull().sum() / len(train)) * 100

# Display the percentages, rounded to two decimal places
print(missing_percentage.round(2))
```

```python
# Count completely duplicated rows in the training dataset
print("Duplicate rows in training data:", train.duplicated().sum())

# Count completely duplicated rows in the test dataset
print("Duplicate rows in test data:", test.duplicated().sum())
```

```python
# Summarize numerical columns in the training dataset
train.describe()
```

## 4. Categorical Data and Unusual Values

In this section, we will examine categorical variables to identify inconsistent labels, understand the distribution of product and outlet categories, and investigate unusual numerical values.

These findings will guide our data-cleaning and preprocessing decisions.

```python
# Count each fat-content category in the training data
print("Training data:")
print(train["Item_Fat_Content"].value_counts())

# Count each fat-content category in the test data
print("\nTest data:")
print(test["Item_Fat_Content"].value_counts())
```

```python
# Count outlet sizes, including missing values
print("Outlet size distribution:")
print(train["Outlet_Size"].value_counts(dropna=False))

# Count product categories
print("\nProduct type distribution:")
print(train["Item_Type"].value_counts())

# Count outlet types
print("\nOutlet type distribution:")
print(train["Outlet_Type"].value_counts())
```

```python
# Count the number of distinct product identifiers
print("Unique products:", train["Item_Identifier"].nunique())

# Count the number of distinct outlets
print("Unique outlets:", train["Outlet_Identifier"].nunique())
```

```python
# Count records with zero item visibility
zero_visibility = (train["Item_Visibility"] == 0).sum()

print("Rows with zero item visibility:", zero_visibility)
print(
    "Percentage with zero visibility:",
    round(zero_visibility / len(train) * 100, 2),
    "%"
)
```
