## 2. Data Cleaning

Data cleaning is the process of identifying and handling missing values, inconsistent categories, and other data-quality issues.

In this notebook, we will:
- Load the original training and test datasets.
- Create copies to protect the original data.
- Standardize inconsistent category labels.
- Handle missing values.
- Verify the cleaned datasets.

The cleaned datasets will be used in the next stages of the project.

```python
# Import pandas to work with tabular data
import pandas as pd

# Load the original training and test CSV files
train = pd.read_csv("../data/raw/Train.csv")
test = pd.read_csv("../data/raw/Test.csv")

# Create copies so the original data stays unchanged
train_clean = train.copy()
test_clean = test.copy()

# Display the dataset sizes
print("Training shape:", train_clean.shape)
print("Test shape:", test_clean.shape)
```

### 2.1 Standardizing Categorical Labels

The `Item_Fat_Content` column contains inconsistent labels, including `LF`, `low fat`, and `reg`.

These labels represent the same categories as `Low Fat` and `Regular`. We will standardize the labels in both datasets to make them consistent.

```python
# Map inconsistent labels to standard labels
mapping = {
    "LF": "Low Fat",
    "low fat": "Low Fat",
    "reg": "Regular"
}

# Apply the mapping to both datasets
train_clean["Item_Fat_Content"] = train_clean["Item_Fat_Content"].replace(mapping)
test_clean["Item_Fat_Content"] = test_clean["Item_Fat_Content"].replace(mapping)

# Display the cleaned categories
print(train_clean["Item_Fat_Content"].value_counts())
```

### 2.2 Handling Missing Values

The initial data inspection identified missing values in `Item_Weight` and `Outlet_Size`.

We will use:
- The median of `Item_Weight` to fill missing item weights.
- The most frequent category of `Outlet_Size` to fill missing outlet sizes.

Both replacement values will be calculated from the training dataset and then applied to the training and test datasets.

```python
# Calculate replacement values from training data
weight_median = train_clean["Item_Weight"].median()
outlet_mode = train_clean["Outlet_Size"].mode()[0]

# Fill missing item weights
train_clean["Item_Weight"] = train_clean["Item_Weight"].fillna(weight_median)
test_clean["Item_Weight"] = test_clean["Item_Weight"].fillna(weight_median)

# Fill missing outlet sizes
train_clean["Outlet_Size"] = train_clean["Outlet_Size"].fillna(outlet_mode)
test_clean["Outlet_Size"] = test_clean["Outlet_Size"].fillna(outlet_mode)

# Display the replacement values
print("Item weight median:", weight_median)
print("Outlet size mode:", outlet_mode)
```

### 2.3 Verifying the Cleaned Data

After cleaning, we will count the remaining missing values in both datasets.

This step checks whether the missing-value treatment was successful. We will also verify that the datasets still have their original numbers of rows and columns.

```python
# Count all remaining missing values
print("Training missing values:", train_clean.isnull().sum().sum())
print("Test missing values:", test_clean.isnull().sum().sum())

# Confirm the dataset sizes
print("Training shape:", train_clean.shape)
print("Test shape:", test_clean.shape)
```

### 2.4 Data Cleaning Summary

In this notebook, we standardized the `Item_Fat_Content` labels and filled missing values in `Item_Weight` and `Outlet_Size`.

We preserved the original datasets and performed cleaning on separate copies. The cleaned datasets are now ready for further investigation and exploratory data analysis.

The chosen imputation methods are baseline decisions and may be improved during model development.
